```python
import pandas as pd
import numpy as np

from sklearn.cluster import KMeans, SpectralClustering, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

from umap import UMAP
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("EE965_data.csv", header=0, index_col=0)
data.dropna(inplace=True)
data_index = data.index
```

```python
# clustering the data
kmeans_pred = KMeans(n_clusters=8, random_state=0).fit_predict(data)
spectral_pred = SpectralClustering(n_clusters=8, affinity='rbf', random_state=0, gamma=0.001).fit_predict(data)
heirar_pred = AgglomerativeClustering(n_clusters=8).fit_predict(data)

#checking the clusters distribution
print("kmeans-predictions")
print(np.unique(kmeans_pred, return_counts=True))
print("SpectralClustering-predictions")
print(np.unique(spectral_pred, return_counts=True))
print("AgglomerativeClustering-predictions")
print(np.unique(heirar_pred, return_counts=True))

# silhouette_scores
kmeans_score = silhouette_score(data, kmeans_pred)
try:
    spectral_score = silhouette_score(data, spectral_pred)
except:
    spectral_score = np.nan
heirar_score = silhouette_score(data, heirar_pred)
print('\n\n\nsilhouette scores')
print('kmeans score ', kmeans_score)
print("spectral score ", spectral_score)
print('AgglomerativeClustering score ', heirar_score)
```

```python
# umap
embedding = UMAP(random_state=0).fit_transform(data)
for val in np.unique(kmeans_pred):
    emb_plot = embedding[kmeans_pred == val,:]
    plt.scatter(emb_plot[:, 0],emb_plot[:, 1], label = val)
plt.legend(loc  ='right', bbox_to_anchor=(1.2,0.5))
plt.title('k-means umap')
plt.show()

for val in np.unique(spectral_pred):
    emb_plot = embedding[spectral_pred == val,:]
    plt.scatter(emb_plot[:, 0],emb_plot[:, 1], label = val)
plt.legend(loc  ='right', bbox_to_anchor=(1.2,0.5))
plt.title('Spectral umap')
plt.show()


for val in np.unique(heirar_pred):
    emb_plot = embedding[heirar_pred == val,:]
    plt.scatter(emb_plot[:, 0],emb_plot[:, 1], label = val)
plt.legend(loc  ='right', bbox_to_anchor=(1.2,0.5))
plt.title('AgglomerativeClustering umap')
plt.show()
```

```python
# PCA

pca = PCA(n_components=10, random_state = 0)
data_scaled = MinMaxScaler().fit_transform(data)
data_pca = pca.fit_transform(data_scaled)
```

```python
# clustering the data
kmeans_pred = KMeans(n_clusters=8, random_state=0).fit_predict(data_pca)
spectral_pred = SpectralClustering(n_clusters=8, affinity='rbf', random_state=0).fit_predict(data_pca)
heirar_pred = AgglomerativeClustering(n_clusters=8).fit_predict(data_pca)

#checking the clusters distribution
print("kmeans-predictions")
print(np.unique(kmeans_pred, return_counts=True))
print("SpectralClustering-predictions")
print(np.unique(spectral_pred, return_counts=True))
print("AgglomerativeClustering-predictions")
print(np.unique(heirar_pred, return_counts=True))

# silhouette_scores
kmeans_score = silhouette_score(data, kmeans_pred)
spectral_score = silhouette_score(data, spectral_pred)
heirar_score = silhouette_score(data, heirar_pred)

print('\n\n\nsilhouette scores')
print('kmeans score ', kmeans_score)
print("spectral score ", spectral_score)
print('AgglomerativeClustering score ', heirar_score)
```

```python
# umap
embedding_pca = UMAP(random_state=0).fit_transform(data_pca)
for val in np.unique(kmeans_pred):
    emb_plot = embedding_pca[kmeans_pred == val,:]
    plt.scatter(emb_plot[:, 0],emb_plot[:, 1], label = val)
plt.legend(loc  ='right', bbox_to_anchor=(1.2,0.5))
plt.title('k-means umap')
plt.show()

for val in np.unique(spectral_pred):
    emb_plot = embedding_pca[spectral_pred == val,:]
    plt.scatter(emb_plot[:, 0],emb_plot[:, 1], label = val)
plt.legend(loc  ='right', bbox_to_anchor=(1.2,0.5))
plt.title('Spectral umap')
plt.show()


for val in np.unique(heirar_pred):
    emb_plot = embedding_pca[heirar_pred == val,:]
    plt.scatter(emb_plot[:, 0],emb_plot[:, 1], label = val)
plt.legend(loc  ='right', bbox_to_anchor=(1.2,0.5))
plt.title('AgglomerativeClustering umap')
plt.show()
```

```python
ground_truth = pd.read_csv("./ground_truth.csv", index_col=0, header=None)[1].values
ground_truth = LabelEncoder().fit_transform(ground_truth)

print("Ground truth")
print(np.unique(ground_truth, return_counts=True))

ground_truth_score = silhouette_score(data, ground_truth)
ground_truth_pca_score = silhouette_score(data_pca, ground_truth)

print('\n\n\nsilhouette scores ground truth')
print('ground_truth_score  ', ground_truth_score)
print("ground_truth_pca_score ", ground_truth_pca_score)
```
